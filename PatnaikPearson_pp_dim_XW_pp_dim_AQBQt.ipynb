# Experiment: $PP(XW)$ vs $PP(A Q B Q^t )$
# Suppose that $X$ is $N \times d$, and $W$ is $d \times d$
# and $X_{ij} \in P(\alpha)$, $W_{ij} \in P(\beta)$ for some alpha, beta
# We consider 2 * 2 cases : ( $X = X_{resid}$ : True / False) * ($W = W_{resid}$ : True / False)
# We have SVDs $X = U_X A V_X^t$, $W = U_W B V_W^t$, with $A$ and $B$ diagonal
# Define $Q = V_X^t U_W$
# Does $PP(XW) = PP(A Q B Q^t )$?
#
# Answer : if at least one of $X$ and $W$ is column-demeaned 
# (i.e. $X = X_{resid}$ or $W = W_{resid}$, or both), 
# then $PP(XW) = PP(A Q B Q^t )$
# If neither $X$ nor $W$ is column-demeaned, then in general $PP(XW) \ne PP(A Q B Q^t )$

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors
from scipy.linalg import qr

import math
import pandas as pd

import PatnaikPearson as pp
import cupy
```

# 1. col_demean_X = False, col_demean_W = False 
# $PP(XW) \ne PP(AQBQ^t)$ 

```python
ff_config = pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt_get_default_config()
ff_config["col_demean_X"] = False
ff_config["col_demean_W"] = False

xx_ff = pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt(ff_config)
```

```python
pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt_plot(xx_ff)
```

# 2. col_demean_X = False, col_demean_W = True 
# $PP(XW) = PP(AQBQ^t)$ 

```python
ft_config = pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt_get_default_config()
ft_config["col_demean_X"] = False
ft_config["col_demean_W"] = True

xx_ft = pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt(ft_config)
```

```python
pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt_plot(xx_ft)
```

```python
pp.experiment_pp_dim_XW_vs_pp_dim_AQBQt_plot(xx_tf)
```
