# Experiment: $(1/m) PP(XW)$ vs $(1/d) PP(X)$, $(1/m) PP(W)$
# Suppose that $X$ is $N \times d$, and $W$ is $d \times m$
# and $X_{ij} \in P(\alpha)$, $W_{ij} \in P(\beta)$ for some $\alpha$, $\beta$
# We consider 2 * 2 cases : ( $X = X_{resid}$ : True / False) * ($W = W_{resid}$ : True / False)
## 1. col_demean_X = False, col_demean_W = False - both lower and upper bounds FAIL
## 2. col_demean_X = False, col_demean_W = True - both lower and upper bounds HOLD
## 3. col_demean_X = True, col_demean_W = False - lower bound FAILS. upper bound HOLDS
## 4. col_demean_X = True, col_demean_W = True - both lower and upper bounds HOLD

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors
from scipy.linalg import qr

import math
import pandas as pd

import PatnaikPearson as pp
import cupy
```

# 1. col_demean_X = False, col_demean_W = False 
# in general, lower and upper bounds fail

```python
ff_config = pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W_get_default_config()
#ff_config["num_iterations"] = 10
ff_config["col_demean_X"] = False
ff_config["col_demean_W"] = False

xx_ff = pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W(ff_config)
```

```python
pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W_plot(xx_ff, "this prefix : ")
```

# 2. col_demean_X = False, col_demean_W = True 
# lower and upper bounds both hold

```python
ft_config = pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W_get_default_config()
ft_config["col_demean_X"] = False
ft_config["col_demean_W"] = True

xx_ft = pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W(ft_config)
```

```python
pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W_plot(xx_ft, "this prefix : ")
```

```python
pp.experiment_pp_dim_XW_pp_dim_X_pp_dim_W_plot(xx_tf, "this prefix : ")
```
